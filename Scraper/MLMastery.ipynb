```python
import os
import subprocess
import time
import random
from bs4 import BeautifulSoup
import pandas as pd
import re
```

```python
df = pd.read_csv('mlmastery_1.csv')
```

```python
df
```

```python
contents=[]
```

```python
for url in df['links']:
    print("Scraping:", url)

    temp_file = "tmp_page.html"

    cmd = [
        "curl.exe",
        "-s", 
        "-L", 
        "-A", "Mozilla/5.0", 
        url,
        "-o", temp_file
    ]
    
    subprocess.run(cmd)

    with open(temp_file, "r", encoding="utf-8") as f:
        html = f.read()

    soup = BeautifulSoup(html, "html.parser")

    content = soup.select("section.entry")
    contents.append(content)

    print(len(contents))

    time.sleep(random.uniform(1.0, 2.0))
```

```python
df['contents']=contents
```
