```python
for i in range(1, 1307):
    url = f"https://lifehacker.com/latest/page/{i}"
    print(f"Scraping: {url} titles: {len(titles)} links: {len(links)} dates: {len(dates)}")

    try:
        driver.get(url)
        time.sleep(random.uniform(2.5, 4.0)) 
        html = driver.page_source

        if "Just a moment" in html:
            print("Cloudflare challenge")
            time.sleep(10)
            html = driver.page_source
            if "Just a moment" in html:
                print("Blocked, skip page")
                continue
        
        soup = BeautifulSoup(html, "html.parser")
        
        found_titles = soup.select("a[data-ga-item='title']")
        found_dates = soup.select("time")
        
        for el in found_titles:
            titles.append(el.get_text(strip=True))
            links.append(el.get("href"))

        for d in found_dates:
            dates.append(d.get_text(strip=True))

    except Exception as e:
        print("Error:", e)
        time.sleep(10)
        continue
```

```python
quarters = [("January","February","March"),("April","May","June"),("July","August","September"),("October","November","December")]
```

```python
def new_driver():
    options = uc.ChromeOptions()
    driver = uc.Chrome(
        use_subprocess=True,
        options=options,
        client_config={
            "connection_timeout": 300,  
            "read_timeout": 300          
        }
    )
    return driver
driver=new_driver()
```

```python
df['year'] = df['dates'].map(lambda x:x[-4:])
```

```python
df.head()
```

```python
new_df = pd.DataFrame(data={"titles":[],"links":[],"dates":[],"content":[],"year":[]},columns=["titles","links","dates","content","year"])
```

```python
for year in ["2025","2024","2023","2022","2021","2020"]:
    for q in [1,2,3,4]:
        cur_year = df[df['year'] == year]
        selected = []
        notselected = set()  

        while len(selected) < 500:
            sel = cur_year.sample(n=1)
            idx = sel.index[0]  

            if idx in notselected:
                continue  
            notselected.add(idx)

            date_str = sel.at[idx, 'dates']  
            if date_str.split()[0] in quarters[q-1]:
                selected.append(sel.iloc[0])  
                
            if len(notselected) >= len(cur_year):
                break  
                
        selected_df = pd.DataFrame(selected)
        new_df = pd.concat([new_df, selected_df], ignore_index=True)
```

```python
def get_quarter_range(month):
    if month in [1,2,3]:
        return [1,2,3]
    if month in [4,5,6]:
        return [4,5,6]
    if month in [7,8,9]:
        return [7,8,9]
    return [10,11,12]
```
