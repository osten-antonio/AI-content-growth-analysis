```python
import pandas as pd
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
from sklearn.linear_model import ElasticNet
```

```python
df = pd.read_csv('aggregate_blog.csv')
```

```python
X = df[['year','month_sin','month_cos','ai_pct_lag1','ai_pct_lag2','ai_pct_ma3','time_index']]
y = df['ai_percentage']
```

```python
tscv = TimeSeriesSplit(n_splits=5)

pipeline = Pipeline([
    ('scaler', StandardScaler()), 
    ('svr', SVR())
])


param_grid = {
    'svr__kernel': ['rbf', 'linear'],
    'svr__C': [0.001, 0.1, 1, 5, 3, 10],
    'svr__gamma': ['scale', 'auto']
}

scoring = {
    'MAE': 'neg_mean_absolute_error',
    'RMSE': 'neg_root_mean_squared_error',
    'R2': 'r2'
}

model = GridSearchCV(
    pipeline,
    param_grid,
    cv=tscv,
    scoring=scoring,
    refit='RMSE',
)

model.fit(X, y)
```

```python
pd.set_option('display.max_columns', None)
pd.DataFrame(model.cv_results_).sort_values('rank_test_RMSE')
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()), 
    ('knn', KNeighborsRegressor())
])


param_grid = {
    'knn__n_neighbors': [5, 7, 9, 11],  
    'knn__weights': ['uniform'],  
    'knn__p': [1, 2],
    'knn__algorithm': ['ball_tree'],  
    'knn__leaf_size': [10,20, 30]  
}

scoring = {
    'MAE': 'neg_mean_absolute_error',
    'RMSE': 'neg_root_mean_squared_error',
    'R2': 'r2'
}

tscv = TimeSeriesSplit(n_splits=5)

model = GridSearchCV(
    pipeline, 
    param_grid,
    cv=tscv,
    scoring=scoring,
    refit='RMSE',
    n_jobs=-1,
    verbose=2,
    return_train_score=True
)

model.fit(X, y)
```

```python
pd.DataFrame(model.cv_results_).sort_values('rank_test_RMSE')
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()), 
    ('rfr', RandomForestRegressor())
])

param_grid= {
    'rfr__n_estimators': [50, 100, 200, 300],
    'rfr__max_depth': [2, 6, 10, 20, None],
    'rfr__min_samples_split': [2, 5],
    'rfr__min_samples_leaf': [1, 2],
    'rfr__max_features': ['sqrt', 'log2',None]
}

model = GridSearchCV(
    pipeline,
    param_grid,
    cv=tscv,
    scoring=scoring,
    refit='RMSE',
)

model.fit(X, y)
```

```python
pd.DataFrame(model.cv_results_).sort_values('rank_test_RMSE')
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()), 
    ('enet', ElasticNet(max_iter=10000))
])

param_grid = {
    'enet__alpha': [0.01,0.1, 1, 10],
    'enet__l1_ratio': [0.1, 0.25, 0.5, 0.75]
}

model = GridSearchCV(
    pipeline,
    param_grid,
    cv=tscv,
    scoring=scoring,
    refit='RMSE'
)

model.fit(X, y)
```

```python
pd.DataFrame(model.cv_results_).sort_values('rank_test_RMSE')
```
