```python
import pandas as pd
import os
import numpy as np

from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import make_scorer, f1_score, roc_auc_score, accuracy_score
from scipy.sparse import load_npz
```

```python
X_train=load_npz("X_train.npz")
X_test=load_npz("X_test.npz")

y_train=np.load("Y_train.npy")
y_test=np.load("Y_test.npy")
```

```python
X_small = X_train[:20000]
y_small = y_train[:20000]

X_small_test = X_test[:20000]
y_small_test = y_test[:20000]
```

```python
scoring = {
    'auc': 'roc_auc',
    'accuracy': 'accuracy',
    'f1': 'f1'
}

param_grid= {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2',None]
}

cv = StratifiedKFold(n_splits=5,shuffle=True,random_state=0)

model = GridSearchCV(
    estimator=RandomForestClassifier(random_state=5,bootstrap=False),
    param_grid=param_grid,
    scoring=scoring,
    refit='f1',  
    cv=cv,
    verbose=1,
    n_jobs=-1
)

model.fit(X_small, y_small)
```

```python
pd.set_option('display.max_columns', None)
pd.DataFrame(model.cv_results_).sort_values('rank_test_f1')
```

```python
best_model = model.best_estimator_

y_pred = best_model.predict(X_small_test)

test_accuracy = accuracy_score(y_small_test, y_pred)
test_f1 = f1_score(y_small_test, y_pred)
test_auc = roc_auc_score(y_small_test, best_model.predict_proba(X_small_test)[:, 1])

print("Test Accuracy:", test_accuracy)
print("Test F1 Score:", test_f1)
print("Test AUC:", test_auc)
```
